```python
#Outlier Detection
numeric_cols = ['Sales', 'Quantity', 'Discount', 'Profit']

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_limit) | (df[col] > upper_limit)]

    print(f"{col}: {len(outliers)} outliers")
```

### Outlier Analysis — Key Observations

- Sales contains 1,167 statistical outliers, indicating the presence of unusually high or low transaction values.
- Quantity contains 170 statistical outliers, representing transactions with unusually high quantities.
- Discount contains 856 statistical outliers, indicating several transactions with unusually high discount levels.
- Profit contains 1,881 statistical outliers, reflecting substantial variation in transaction profitability.
- These outliers are not automatically considered data errors and will be investigated further using business context before deciding whether any treatment is required.

## Discount Level Analysis

Discount levels are grouped into ranges to examine how profitability changes across different discount levels. This helps identify discount levels associated with higher or lower average profit.

```python
df['Discount_Group'] = pd.cut(
    df['Discount'],
    bins=[-0.01, 0, 0.10, 0.20, 0.30, 0.40, 0.50, 0.60, 0.80],
    labels=['0%', '1-10%', '11-20%', '21-30%', '31-40%',
            '41-50%', '51-60%', '61-80%']
)

df.groupby('Discount_Group', observed=True)['Profit'].mean()
```

"Higher discount levels are associated with lower average profitability in the dataset."

### Interpretation

Profitability varies substantially across discount levels. Average Profit is positive at lower discount levels, with the highest average Profit observed in the 1–10% discount group (96.06). Ave
rage Profit declines as discount levels increase and becomes negative in the 21–30% range. The 41–50% discount group shows the lowest average Profit (-298.70). Overall, higher discount levels are associated with lower average profitability, although this analysis does not establish a causal relationship.



Business Insight

Lower discount levels are associated with better profitability, while higher discount levels are associated with negative average Profit. This suggests that discounting should be carefully managed to avoid reducing profitability.

# **Profitability Analysis**

```python
df['Profit_Margin'] = (df['Profit'] / df['Sales']) * 100

df[['Sales', 'Profit', 'Profit_Margin']].head()
```

### Interpretation

Profit Margin varies considerably across individual records. Some transactions generate positive margins, while others have negative margins, indicating that they generate losses despite having positive Sales. This demonstrates that Sales alone cannot be used to evaluate profitability and that Profit Margin provides additional insight into the efficiency of revenue generation.

```python
df['Profit_Margin'].describe()
```

### Interpretation

Profit Margin shows substantial variation across the 9,994 records. The mean Profit Margin is 12.03%, while the median is 27%, indicating that extreme negative-margin records are pulling the average downward. Profit Margin ranges from -275% to 50%, showing that some transactions generate substantial losses while others achieve relatively high margins. The high standard deviation of 46.68% further indicates significant variation in transaction-level profitability.

Now let's calculate the overall Profit Margin for the entire business:

```python
overall_profit_margin = (df['Profit'].sum() / df['Sales'].sum()) * 100

overall_profit_margin
```

The Superstore dataset generates an overall profit margin of approximately 12.47%, indicating that the business retains about $12.47 in profit for every $100 in Sales.
