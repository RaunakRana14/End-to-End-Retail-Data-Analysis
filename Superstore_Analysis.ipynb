## Pareto Analysis of Product Profitability

Pareto analysis is used to determine whether a relatively small group of products contributes a large proportion of the total Profit.

```python
product_profit = (
    df.groupby('Product Name')['Profit']
      .sum()
      .sort_values(ascending=False)
)

product_profit.head(10)
```

### Cumulative Profit Contribution

Cumulative Profit and cumulative Profit percentage are calculated to determine how quickly the highest-profit products contribute to the overall Profit.

```python
pareto_df = product_profit.to_frame(name='Profit')

pareto_df['Cumulative_Profit'] = pareto_df['Profit'].cumsum()

total_profit = product_profit.sum()

pareto_df['Cumulative_Profit_%'] = (
    pareto_df['Cumulative_Profit'] / total_profit
) * 100

pareto_df.head(10)
```

```python
positive_product_profit = product_profit[product_profit > 0].sort_values(ascending=False)

positive_product_profit.head(10)
```

```python
pareto_positive = positive_product_profit.to_frame(name='Profit')

pareto_positive['Cumulative_Profit'] = pareto_positive['Profit'].cumsum()

total_positive_profit = positive_product_profit.sum()

pareto_positive['Cumulative_Profit_%'] = (
    pareto_positive['Cumulative_Profit'] /
    total_positive_profit
) * 100

pareto_positive.head(10)
```

```python
products_for_80_profit = (
    pareto_positive['Cumulative_Profit_%'].ge(80).argmax() + 1
)

products_for_80_profit
```

```python
total_profitable_products = len(pareto_positive)

percentage_products = (
    products_for_80_profit / total_profitable_products
) * 100

percentage_products
```

### Pareto Analysis — Interpretation

Pareto analysis shows that approximately 22.11% of profitable products contribute 80% of the total positive Profit. This indicates a strong concentration of profitability among a relatively small proportion of products. These high-contributing products should receive greater attention for inventory planning, sales strategy, and profitability management.

## Advanced Analysis — RFM Customer Analysis

RFM analysis is used to evaluate customers based on Recency, Frequency, and Monetary value to identify high-value and less-engaged customers.

```python
df['Order Date'].max()
```

```python
reference_date = df['Order Date'].max() + pd.Timedelta(days=1)

recency = (
    df.groupby('Customer ID')['Order Date']
      .max()
      .apply(lambda x: (reference_date - x).days)
)

recency.head()
```

Lower Recency = better customer engagement.



```python
frequency = (
    df.groupby('Customer ID')['Order ID']
      .nunique()
)

frequency.head()
```

```python
monetary = (
    df.groupby('Customer ID')['Sales']
      .sum()
)

monetary.head()
```

```python
rfm = pd.concat(
    [recency, frequency, monetary],
    axis=1
)

rfm.columns = ['Recency', 'Frequency', 'Monetary']

rfm.head()
```

```python
rfm['R_Score'] = pd.qcut(
    rfm['Recency'],
    5,
    labels=[5, 4, 3, 2, 1]
)

rfm['F_Score'] = pd.qcut(
    rfm['Frequency'].rank(method='first'),
    5,
    labels=[1, 2, 3, 4, 5]
)

rfm['M_Score'] = pd.qcut(
    rfm['Monetary'],
    5,
    labels=[1, 2, 3, 4, 5]
)

rfm.head()
```

```python
rfm['RFM_Score'] = (
    rfm['R_Score'].astype(int) +
    rfm['F_Score'].astype(int) +
    rfm['M_Score'].astype(int)
)

rfm.head()
```

```python
rfm['RFM_Score'].value_counts().sort_index()
```

```python
rfm['RFM_Score'].describe()
```

### RFM Score Interpretation

The RFM score ranges from 3 to 15, with an average and median score of approximately 9. Most customers fall within the middle range of 7 to 11. This indicates that the customer base is largely concentrated around moderate levels of recency, frequency, and monetary value, while customers with extremely high or low overall RFM scores represent smaller portions of the customer base.

```python
def rfm_segment(row):
    if row['R_Score'] >= 4 and row['F_Score'] >= 4 and row['M_Score'] >= 4:
        return 'High Value'
    elif row['R_Score'] <= 2 and (row['F_Score'] >= 3 or row['M_Score'] >= 4):
        return 'At Risk'
    elif row['F_Score'] >= 4:
        return 'Loyal'
    elif row['R_Score'] <= 2 and row['F_Score'] <= 2 and row['M_Score'] <= 2:
        return 'Low Value'
    else:
        return 'Regular'

rfm['Segment'] = rfm.apply(rfm_segment, axis=1)

rfm['Segment'].value_counts()

```

Main insights

1. Regular customers are the largest group — 272 (34.3%)

These customers have moderate overall engagement and represent the largest opportunity for conversion into Loyal/High Value customers.

2. At Risk customers are significant — 167 (21.1%)

This is an important finding. More than 1 in 5 customers fall into the At Risk segment according to our RFM rules.

These customers deserve retention/re-engagement strategies.

3. High Value customers = 106 (13.4%)

These are the customers with strong Recency, Frequency and Monetary scores. They are the most valuable group to retain.

4. Low Value = 117 (14.8%)

These customers have relatively low engagement and spending. They can be targeted with lower-cost marketing rather than expensive retention efforts.

```python
rfm_analysis = rfm.reset_index()

customer_metrics = (
    df.groupby('Customer ID')
      .agg(
          Sales=('Sales', 'sum'),
          Profit=('Profit', 'sum')
      )
)

rfm_analysis = rfm_analysis.merge(
    customer_metrics,
    on='Customer ID'
)

rfm_analysis['Profit_Margin'] = (
    rfm_analysis['Profit'] /
    rfm_analysis['Sales']
) * 100

segment_performance = (
    rfm_analysis.groupby('Segment')
    .agg(
        Customers=('Customer ID', 'count'),
        Sales=('Sales', 'sum'),
        Profit=('Profit', 'sum'),
        Profit_Margin=('Profit_Margin', 'mean')
    )
    .sort_values('Profit', ascending=False)
)

segment_performance
```

```python
# ============================================
# ADVANCED ANALYSIS — PRODUCT PROFITABILITY
# ============================================

# 1. Product-level Sales, Profit and Quantity
product_analysis = (
    df.groupby('Product Name')
      .agg(
          Sales=('Sales', 'sum'),
          Profit=('Profit', 'sum'),
          Quantity=('Quantity', 'sum'),
          Orders=('Order ID', 'nunique')
      )
)

# 2. Product-level Profit Margin
product_analysis['Profit_Margin'] = (
    product_analysis['Profit'] /
    product_analysis['Sales']
) * 100


# ============================================
# 3. High Sales + High Profit
# ============================================

sales_median = product_analysis['Sales'].median()
profit_median = product_analysis['Profit'].median()

product_analysis['Performance'] = 'Low Sales - Low Profit'

product_analysis.loc[
    (product_analysis['Sales'] >= sales_median) &
    (product_analysis['Profit'] >= profit_median),
    'Performance'
] = 'High Sales - High Profit'

product_analysis.loc[
    (product_analysis['Sales'] >= sales_median) &
    (product_analysis['Profit'] < profit_median),
    'Performance'
] = 'High Sales - Low Profit'

product_analysis.loc[
    (product_analysis['Sales'] < sales_median) &
    (product_analysis['Profit'] >= profit_median),
    'Performance'
] = 'Low Sales - High Profit'


# ============================================
# 4. Top 10 Products by Sales
# ============================================

top_sales_products = (
    product_analysis
    .sort_values('Sales', ascending=False)
    .head(10)
)


# ============================================
# 5. Top 10 Products by Profit
# ============================================

top_profit_products = (
    product_analysis
    .sort_values('Profit', ascending=False)
    .head(10)
)


# ============================================
# 6. Bottom 10 Products by Profit
# ============================================

bottom_profit_products = (
    product_analysis
    .sort_values('Profit', ascending=True)
    .head(10)
)


# ============================================
# 7. High Sales but Low/Negative Profit
# ============================================

high_sales_low_profit = (
    product_analysis[
        (product_analysis['Sales'] >= sales_median) &
        (product_analysis['Profit'] < 0)
    ]
    .sort_values('Sales', ascending=False)
)


# ============================================
# 8. Low Sales but High Profit
# ============================================

low_sales_high_profit = (
    product_analysis[
        (product_analysis['Sales'] < sales_median) &
        (product_analysis['Profit'] > 0)
    ]
    .sort_values('Profit', ascending=False)
)


# ============================================
# 9. Product Performance Distribution
# ============================================

performance_summary = (
    product_analysis['Performance']
    .value_counts()
    .to_frame('Product_Count')
)


# ============================================
# 10. Display Main Results
# ============================================

print("Top 10 Products by Sales:")
display(top_sales_products)

print("\nTop 10 Products by Profit:")
display(top_profit_products)

print("\nBottom 10 Products by Profit:")
display(bottom_profit_products)

print("\nHigh Sales but Negative Profit:")
display(high_sales_low_profit)

print("\nLow Sales but High Profit:")
display(low_sales_high_profit)

print("\nProduct Performance Distribution:")
display(performance_summary)
```

```python


from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Features and target
X = df[
    [
        'Sales',
        'Quantity',
        'Discount',
        'Category',
        'Sub-Category',
        'Region',
        'Segment'
    ]
]

y = df['Profit']


# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)


# Categorical and numerical columns
categorical_features = [
    'Category',
    'Sub-Category',
    'Region',
    'Segment'
]

numerical_features = [
    'Sales',
    'Quantity',
    'Discount'
]


# Preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        (
            'cat',
            OneHotEncoder(handle_unknown='ignore'),
            categorical_features
        ),
        (
            'num',
            'passthrough',
            numerical_features
        )
    ]
)


# Random Forest model
model = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        (
            'regressor',
            RandomForestRegressor(
                n_estimators=200,
                random_state=42,
                n_jobs=-1
            )
        )
    ]
)


# Train model
model.fit(X_train, y_train)


# Predictions
y_pred = model.predict(X_test)


# Evaluation
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("MAE:", mae)
print("RMSE:", rmse)
print("R² Score:", r2)
```

```python

# ML FEATURE ENGINEERING


# Make a copy
ml_df = df.copy()


# 1. Date Features


ml_df['Order Year'] = ml_df['Order Date'].dt.year
ml_df['Order Month'] = ml_df['Order Date'].dt.month
ml_df['Order Quarter'] = ml_df['Order Date'].dt.quarter


# 2. Shipping Time

ml_df['Shipping Days'] = (
    ml_df['Ship Date'] - ml_df['Order Date']
).dt.days

# 3. Product-level historical features


product_avg_profit = (
    ml_df.groupby('Product Name')['Profit']
         .mean()
         .rename('Product_Avg_Profit')
)

product_avg_sales = (
    ml_df.groupby('Product Name')['Sales']
         .mean()
         .rename('Product_Avg_Sales')
)

ml_df = ml_df.merge(
    product_avg_profit,
    on='Product Name',
    how='left'
)

ml_df = ml_df.merge(
    product_avg_sales,
    on='Product Name',
    how='left'
)

# 4. Customer-level historical features


customer_avg_sales = (
    ml_df.groupby('Customer ID')['Sales']
         .mean()
         .rename('Customer_Avg_Sales')
)

customer_avg_profit = (
    ml_df.groupby('Customer ID')['Profit']
         .mean()
         .rename('Customer_Avg_Profit')
)

ml_df = ml_df.merge(
    customer_avg_sales,
    on='Customer ID',
    how='left'
)

ml_df = ml_df.merge(
    customer_avg_profit,
    on='Customer ID',
    how='left'
)

# 5. Select ML features


features = [
    'Sales',
    'Quantity',
    'Discount',
    'Category',
    'Sub-Category',
    'Region',
    'Segment',
    'Order Year',
    'Order Month',
    'Order Quarter',
    'Shipping Days',
    'Product_Avg_Profit',
    'Product_Avg_Sales',
    'Customer_Avg_Sales'
]

X = ml_df[features]
y = ml_df['Profit']



# 6. Check the dataset

print("Feature shape:", X.shape)
print("\nFeatures:")
print(X.columns.tolist())

print("\nMissing values:")
print(X.isnull().sum())
```

```python

# ML MODEL COMPARISON


from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

import numpy as np
import pandas as pd


# 1. Train-Test Split


X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)


# 2. Feature Types


categorical_features = [
    'Category',
    'Sub-Category',
    'Region',
    'Segment'
]

numerical_features = [
    'Sales',
    'Quantity',
    'Discount',
    'Order Year',
    'Order Month',
    'Order Quarter',
    'Shipping Days',
    'Product_Avg_Profit',
    'Product_Avg_Sales',
    'Customer_Avg_Sales'
]


# 3. Preprocessing


preprocessor = ColumnTransformer(
    transformers=[
        (
            'cat',
            OneHotEncoder(handle_unknown='ignore'),
            categorical_features
        ),
        (
            'num',
            'passthrough',
            numerical_features
        )
    ]
)

# 4. Models


models = {

    'Linear Regression': LinearRegression(),

    'Random Forest': RandomForestRegressor(
        n_estimators=200,
        random_state=42,
        n_jobs=-1
    ),

    'Gradient Boosting': GradientBoostingRegressor(
        n_estimators=200,
        learning_rate=0.05,
        max_depth=3,
        random_state=42
    )
}

# 5. Train + Evaluate


results = []

for name, regressor in models.items():

    pipeline = Pipeline(
        steps=[
            ('preprocessor', preprocessor),
            ('regressor', regressor)
        ]
    )

    pipeline.fit(X_train, y_train)

    predictions = pipeline.predict(X_test)

    mae = mean_absolute_error(
        y_test,
        predictions
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_test,
            predictions
        )
    )

    r2 = r2_score(
        y_test,
        predictions
    )

    results.append({
        'Model': name,
        'MAE': mae,
        'RMSE': rmse,
        'R2_Score': r2
    })


# 6. Comparison Table


model_results = pd.DataFrame(results)

model_results = model_results.sort_values(
    'R2_Score',
    ascending=False
)

model_results
```
